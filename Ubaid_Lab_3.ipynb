```python
from google import genai
```

```python
client = genai.Client(api_key="")
```

```python
MODEL = "models/gemini-3.5-flash"
```

```python
def travel_agent(user_input):

    prompt = f"""
You are a professional travel agent for Karachi tourism.

Provide a well-structured travel plan for Karachi with clear headings:

1. Destination Overview
2. Budget Travel Options
3. Places to Visit
4. Food Suggestions
5. Money-Saving Tips

Make the response detailed but easy for students.

User request: {user_input}

Karachi Travel Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text
```

```python
print(travel_agent("I want to visit khi for 3 days"))
```

```python
print("Gemini Travel Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = travel_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
