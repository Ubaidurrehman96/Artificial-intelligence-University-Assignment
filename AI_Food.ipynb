```python
from google import genai
```

```python
client = genai.Client(api_key="")
```

```python
MODEL = "models/gemini-3.5-flash"
```

```python
def food_agent(user_input):

    prompt = f"""
You are a helpful food recommendation agent for Karachi.

Recommend restaurants near Karachi based on the user's request.

Provide:

1. Restaurant Name
2. Location
3. Recommended Food
4. Approximate Price in PKR
5. Short Reason to Try

Keep the answer simple and easy for students.

User request: {user_input}

Food Recommendations:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text
```

```python
print(food_agent("Suggest affordable restaurants near Karachi"))
```
